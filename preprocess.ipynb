First, we import packages.

```python
import pandas as pd
from pathlib import Path
pd.set_option('future.no_silent_downcasting', True)
```

Then, we provide the data from .csv to pandas dataframes (remember to first download the files into data/raw folder and unzip it there).

```python
PROJECT_ROOT = Path.cwd()
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
```

```python
df_train = pd.read_csv(RAW_DATA_DIR / "train.csv")
df_test = pd.read_csv(RAW_DATA_DIR / "test.csv")
df_train
```

```python
df_train.columns.values
```

Let's find out what values does each column contain.

```python
for column in df_train.columns.values:
    print(f"Unique values in column {column} are: {df_train[column].unique()}")
```

First, we want to take the number of people in each group ID

```python
for df in [df_train, df_test]:
    df['NumberGroup'] = df['PassengerId'].str.split('_').str[1]
    df['NumberGroup'] = df['NumberGroup'].astype(int)
```

```python
df_train
```

Now we want to fill missing values and perform one-hot-encoding on column HomePlanet.

```python
def encode_homeplanet(df):
    df = df.copy()
    df['HomePlanet'] = df['HomePlanet'].fillna('MissingHP')
    one_hot = pd.get_dummies(df['HomePlanet'], prefix='HomePlanet')
    df = df.drop('HomePlanet', axis=1).join(one_hot)
    return df

df_train = encode_homeplanet(df_train)
df_test  = encode_homeplanet(df_test)
```

```python
df_train
```

We want to handle missing values in CryoSleep column.

```python
for df in [df_train, df_test]:
    df['CryoSleep'] = df['CryoSleep'].fillna(df['CryoSleep'].mode()[0])
```

```python
df_train
```

Now we want to transform the Cabin column into three different columns.

```python
def encode_cabin(df):
    df = df.copy()
    df['CabinDeck'] = df['Cabin'].str.split('/').str[0]
    df['CabinNum'] = df['Cabin'].str.split('/').str[1]
    df['CabinSide'] = df['Cabin'].str.split('/').str[2]

    df['CabinDeck'] = df['CabinDeck'].fillna(df['CabinDeck'].mode()[0])
    df['CabinNum'] = df['CabinNum'].fillna(df['CabinNum'].mode()[0]).astype(int)
    df['CabinSide'] = df['CabinSide'].fillna(df['CabinSide'].mode()[0])

    one_hot1 = pd.get_dummies(df['CabinDeck'], prefix='CabinDeck')
    df = df.drop('CabinDeck', axis=1).join(one_hot1)

    one_hot2 = pd.get_dummies(df['CabinSide'], prefix='CabinSide')
    df = df.drop('CabinSide', axis=1).join(one_hot2)

    df = df.drop('Cabin', axis=1)
    return df

df_train = encode_cabin(df_train)
df_test = encode_cabin(df_test)
```

```python
df_train
```

Now we perform one-hot-encoding on Destination.

```python
def encode_destination(df):
    df = df.copy()

    df['Destination'] = df['Destination'].fillna(df['Destination'].mode()[0])

    one_hot = pd.get_dummies(df['Destination'], prefix='Destination')
    df = df.drop('Destination', axis=1).join(one_hot)
    return df

df_train = encode_destination(df_train)
df_test  = encode_destination(df_test)
```

```python
df_train
```

Now we will handle with missing values in other columns.

```python
def missing_rest_values(df):
    df = df.copy()

    df['Age'] = df['Age'].fillna(df['Age'].mean())
    df['VIP'] = df['VIP'].fillna(df['VIP'].mode()[0])

    for column in ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']:
        df[column] = df[column].fillna(df[column].mean())

    return df

df_train = missing_rest_values(df_train)
df_test = missing_rest_values(df_test)
```

```python
df_train
```

Now we will handle with Name column - we will count how many names and surnames were encountered.

```python
def names_handling(df):
    df = df.copy()
    df['Name'] = df['Name'].fillna('Missing Missing')
    df[['FirstName', 'LastName']] = df['Name'].str.split(' ', n=1, expand=True)
    return df

df_train = names_handling(df_train)
df_test = names_handling(df_test)
df_train
```

```python
first_name_counts = df_train['FirstName'].value_counts()
last_name_counts  = df_train['LastName'].value_counts()

first_name_mean = first_name_counts.drop('Missing', errors='ignore').mean().round()
last_name_mean  = last_name_counts.drop('Missing', errors='ignore').mean().round()
```

```python
def names_counting(df):
    df = df.copy()

    df['FirstNamesCount'] = df['FirstName'].map(first_name_counts).fillna(1)
    df['LastNamesCount']  = df['LastName'].map(last_name_counts).fillna(1)

    df.loc[df['FirstName'] == 'Missing', 'FirstNamesCount'] = first_name_mean
    df.loc[df['LastName'] == 'Missing', 'LastNamesCount'] = last_name_mean

    df = df.drop(['FirstName', 'LastName', 'Name'], axis=1)
    return df

df_train = names_counting(df_train)
df_test = names_counting(df_test)
```

```python
target = 'Transported'

X_train = df_train.drop(columns=[target])
y_train = df_train[target]

X_test = df_test.copy()
```
